```python
# Build the final monthly search dataset
con.execute(f"""
    CREATE OR REPLACE VIEW monthly_search AS
    SELECT
        client_hash_id,
        content_hash_id,
        DATE_TRUNC('month', report_date) AS month,
        SUM(gsc_impressions) AS impressions,
        SUM(gsc_clicks) AS clicks,
        CASE
            WHEN SUM(gsc_impressions) > 0
            THEN SUM(gsc_clicks) * 100.0 / SUM(gsc_impressions)
            ELSE NULL
        END AS ctr,
        AVG(
            CASE
                WHEN gsc_avg_position > 0
                THEN gsc_avg_position
                ELSE NULL
            END
        ) AS avg_position,
        COUNT(DISTINCT report_date) AS observed_days
    FROM read_parquet('{PERF}')
    WHERE report_date BETWEEN DATE '2026-04-01' AND DATE '2026-06-30'
      AND gsc_data_available = TRUE
    GROUP BY 1, 2, 3
""")

print("Monthly search aggregation created.")
```

```python
# Create the compact April-June feature table
feature_base = con.sql("""
    SELECT
        client_hash_id,
        content_hash_id,

        MAX(CASE WHEN month = DATE '2026-04-01'
            THEN impressions END) AS impressions_apr,

        MAX(CASE WHEN month = DATE '2026-04-01'
            THEN clicks END) AS clicks_apr,

        MAX(CASE WHEN month = DATE '2026-04-01'
            THEN ctr END) AS ctr_apr,

        MAX(CASE WHEN month = DATE '2026-04-01'
            THEN avg_position END) AS position_apr,

        MAX(CASE WHEN month = DATE '2026-06-01'
            THEN impressions END) AS impressions_jun,

        MAX(CASE WHEN month = DATE '2026-06-01'
            THEN clicks END) AS clicks_jun,

        MAX(CASE WHEN month = DATE '2026-06-01'
            THEN ctr END) AS ctr_jun,

        MAX(CASE WHEN month = DATE '2026-06-01'
            THEN avg_position END) AS position_jun

    FROM monthly_search
    GROUP BY 1, 2
""").df()

print("Feature rows:", len(feature_base))
```

```python

# Check period coverage
print({
    "total_items": len(feature_base),
    "april": feature_base["impressions_apr"].notna().sum(),
    "june": feature_base["impressions_jun"].notna().sum(),
    "both": (
        feature_base["impressions_apr"].notna()
        & feature_base["impressions_jun"].notna()
    ).sum()
})
```

```python
# Join search performance with content metadata
feature_df = con.sql(f"""
    SELECT
        f.*,
        c.content_type,
        c.search_volume,
        c.competition,
        c.competition_level,
        c.main_intent,
        c.backlinks,
        c.char_count,
        c.word_count,
        c.content_created_date,
        c.content_updated_date,
        c.last_optimized_date,
        c.is_published,
        c.is_deleted
    FROM feature_base f
    LEFT JOIN read_parquet('{CONTENT}') c
        ON f.content_hash_id = c.content_hash_id
    WHERE f.impressions_apr IS NOT NULL
      AND f.impressions_jun IS NOT NULL
""").df()

print("Eligible feature rows:", len(feature_df))
print(feature_df.head())
```

```python
# Create leakage-safe modeling features
model_features = model_df.copy()

model_features["content_age_days"] = (
    pd.Timestamp("2026-06-30")
    - pd.to_datetime(model_features["content_created_date"])
).dt.days

model_features["days_since_update"] = (
    pd.Timestamp("2026-06-30")
    - pd.to_datetime(model_features["content_updated_date"])
).dt.days

model_features["apr_ctr"] = model_features["ctr_apr"]

model_features["apr_position"] = model_features["position_apr"]

feature_cols = [
    "impressions_apr",
    "clicks_apr",
    "apr_ctr",
    "apr_position",
    "search_volume",
    "competition",
    "backlinks",
    "char_count",
    "word_count",
    "content_age_days",
    "days_since_update"
]

X = model_features[feature_cols].copy()
y = model_features["is_declining"].copy()

print("Features:", X.shape)
print("Target:", y.shape)
print("Positive rate:", y.mean())
```

```python
# Track missing metadata explicitly
for col in [
    "search_volume",
    "competition",
    "backlinks",
    "char_count",
    "word_count",
    "days_since_update"
]:
    X[f"{col}_missing"] = X[col].isna().astype(int)

print("Feature columns:", len(X.columns))
```

```python
# Build a simple refresh opportunity baseline
baseline_df = model_features[
    [
        "client_hash_id",
        "content_hash_id",
        "impressions_apr",
        "apr_ctr",
        "days_since_update",
        "is_declining"
    ]
].copy()

baseline_df["volume_rank"] = baseline_df["impressions_apr"].rank(pct=True)
baseline_df["ctr_opportunity"] = 1 - baseline_df["apr_ctr"].rank(pct=True)
baseline_df["freshness_opportunity"] = baseline_df["days_since_update"].rank(pct=True)

baseline_df["baseline_score"] = (
    0.40 * baseline_df["volume_rank"]
    + 0.35 * baseline_df["ctr_opportunity"]
    + 0.25 * baseline_df["freshness_opportunity"]
)

baseline_df["reason_code"] = "HIGH_VOLUME_LOW_CTR_OLDER"

baseline_df = baseline_df.sort_values(
    "baseline_score",
    ascending=False
).reset_index(drop=True)

print(baseline_df.head(10))
```

```python
# Measure baseline precision at the top 50
top50_baseline = baseline_df.head(50)

baseline_precision_50 = top50_baseline["is_declining"].mean()

print("Baseline Precision@50:", round(baseline_precision_50, 4))
print("Declining items in top 50:", top50_baseline["is_declining"].sum())
```

```python
# Create a client-grouped train and holdout split
from sklearn.model_selection import GroupShuffleSplit

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, holdout_idx = next(
    splitter.split(
        X,
        y,
        groups=model_features["client_hash_id"]
    )
)

X_train = X.iloc[train_idx].copy()
X_holdout = X.iloc[holdout_idx].copy()

y_train = y.iloc[train_idx].copy()
y_holdout = y.iloc[holdout_idx].copy()

print("Train rows:", len(X_train))
print("Holdout rows:", len(X_holdout))

print(
    "Train clients:",
    model_features.iloc[train_idx]["client_hash_id"].nunique()
)

print(
    "Holdout clients:",
    model_features.iloc[holdout_idx]["client_hash_id"].nunique()
)
```

```python
# Confirm clients do not overlap between train and holdout
train_clients = set(
    model_features.iloc[train_idx]["client_hash_id"]
)

holdout_clients = set(
    model_features.iloc[holdout_idx]["client_hash_id"]
)

overlap = train_clients.intersection(holdout_clients)

print("Client overlap:", len(overlap))
```

```python
# Train a logistic regression model
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

model.fit(X_train, y_train)

holdout_prob = model.predict_proba(X_holdout)[:, 1]

print("Model trained.")
```

```python
# Evaluate the model at the top 50
model_results = model_features.iloc[
    holdout_idx
][["client_hash_id", "content_hash_id"]].copy()

model_results["actual"] = y_holdout.to_numpy()
model_results["score"] = holdout_prob

model_results = model_results.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

model_precision_50 = model_results.head(50)["actual"].mean()

print("ML Precision@50:", round(model_precision_50, 4))
print("Declining items in ML top 50:", model_results.head(50)["actual"].sum())
```

```python
# Compare baseline and ML on the same holdout

# Rebuild baseline ranking for the holdout
holdout_ids = model_features.iloc[
    holdout_idx
][["client_hash_id", "content_hash_id"]]

baseline_holdout = baseline_df.merge(
    holdout_ids,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

baseline_holdout = baseline_holdout.sort_values(
    "baseline_score",
    ascending=False
).reset_index(drop=True)

baseline_precision_50_holdout = (
    baseline_holdout.head(50)["is_declining"].mean()
)

# Train the model
model.fit(X_train, y_train)

holdout_prob = model.predict_proba(X_holdout)[:, 1]

model_results = model_features.iloc[
    holdout_idx
][["client_hash_id", "content_hash_id"]].copy()

model_results["actual"] = y_holdout.to_numpy()
model_results["score"] = holdout_prob

model_results = model_results.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

model_precision_50 = (
    model_results.head(50)["actual"].mean()
)

# Final comparison
print("Baseline Precision@50:", round(baseline_precision_50_holdout, 4))
print("ML Precision@50:", round(model_precision_50, 4))
print(
    "Improvement:",
    round(model_precision_50 - baseline_precision_50_holdout, 4)
)
```

```python
# Train a Random Forest model
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1,
        min_samples_leaf=5
    ))
])

rf_model.fit(X_train, y_train)

rf_prob = rf_model.predict_proba(X_holdout)[:, 1]

rf_results = model_features.iloc[
    holdout_idx
][["client_hash_id", "content_hash_id"]].copy()

rf_results["actual"] = y_holdout.to_numpy()
rf_results["score"] = rf_prob

rf_results = rf_results.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

rf_precision_50 = rf_results.head(50)["actual"].mean()

print("Random Forest Precision@50:", round(rf_precision_50, 4))
print(
    "Declining items in RF top 50:",
    int(rf_results.head(50)["actual"].sum())
)
```

```python
# Compare all approaches
results_table = pd.DataFrame({
    "Approach": [
        "Rule baseline",
        "Logistic Regression",
        "Random Forest"
    ],
    "Precision@50": [
        baseline_precision_50_holdout,
        model_precision_50,
        rf_precision_50
    ]
})

print(results_table)
```

```python
# Build the final ranked refresh queue
recommendations = model_features[
    [
        "client_hash_id",
        "content_hash_id",
        "impressions_apr",
        "apr_ctr",
        "days_since_update",
        "search_volume",
        "impression_change_pct",
        "is_declining"
    ]
].copy()

recommendations["opportunity_score"] = (
    0.40 * recommendations["impressions_apr"].rank(pct=True)
    + 0.35 * (1 - recommendations["apr_ctr"].rank(pct=True))
    + 0.25 * recommendations["days_since_update"].rank(pct=True)
)

recommendations["priority"] = pd.cut(
    recommendations["opportunity_score"],
    bins=[-np.inf, 0.50, 0.75, np.inf],
    labels=["Monitor", "Review", "High Priority"]
)

recommendations["reason_code"] = np.select(
    [
        recommendations["impression_change_pct"] <= -50,
        recommendations["apr_ctr"] <= recommendations["apr_ctr"].median(),
        recommendations["days_since_update"] >= recommendations["days_since_update"].median()
    ],
    [
        "STRONG_DECLINE",
        "LOW_CTR_OPPORTUNITY",
        "OLDER_CONTENT"
    ],
    default="MIXED_SIGNAL"
)

recommendations = recommendations.sort_values(
    "opportunity_score",
    ascending=False
).reset_index(drop=True)

print(recommendations.head(20))
```

```python
# Create a public-safe recommendation table
paper_recommendations = recommendations.head(20).copy()

paper_recommendations["rank"] = range(1, 21)

paper_recommendations = paper_recommendations[
    [
        "rank",
        "priority",
        "reason_code",
        "opportunity_score",
        "impressions_apr",
        "apr_ctr",
        "days_since_update",
        "impression_change_pct"
    ]
]

print(paper_recommendations)
```

## 1. Question

*The research question and the decision it supports.*

## 1. Question

### Research question

Which published content items should be prioritized for refresh review based on observed search-performance signals and content characteristics?

### Decision supported

This work supports the decision of which content items an editorial or SEO team should review first. The output is a ranked decision-support queue that helps focus limited review time on items showing stronger refresh opportunity signals.

The system is intended to support human review. It does not automatically decide that a page should be rewritten, and it does not claim to predict Google's ranking algorithm or prove that a content refresh will recover search performance.

### Expected action

Items near the top of the ranked queue should receive earlier human review. Reviewers can then determine whether the observed decline is meaningful, whether the page is still strategically important, and whether a refresh, metadata review, monitoring, or no action is appropriate.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

## 2. Data

### Dataset

This analysis uses the FlyRank internship warehouse release `v20260703`, hosted on Hugging Face.

The main performance source is:

- `fact_content_daily_performance`
- `dim_content`

The daily performance table contains search and analytics measurements at the report-date, client, and content level. The content dimension provides metadata such as search volume, competition, content type, word count, backlinks, and content dates.

### Time window

The main analysis window is April 1, 2026 through June 30, 2026.

April is used as the historical comparison period and June as the later observed outcome period. Only content with observations in both periods is included in the comparison population.

### Initial warehouse checks

The April-June performance window contained 33,806,178 daily rows across 66 clients and 409,326 content items.

A separate June development sample contained 11,694,072 rows. After checking the expected daily grain, 11,687,682 distinct `report_date × client_hash_id × content_hash_id` records remained. Duplicate rows were exact duplicates and were removed before aggregation.

The final comparison population contained 158,700 client-content pairs with both April and June GSC observations.

### Modeling population

The final modeling population was restricted to:

1. Published content.
2. Non-deleted content.
3. At least 100 April GSC impressions.
4. Observations available in both April and June.

This produced 104,301 modeling records.

The minimum April-impression threshold was used to reduce instability from very small percentage-change denominators.

### Features used

Historical features available before the June outcome included:

- April impressions
- April clicks
- April CTR
- April average position
- Search volume
- Competition
- Backlinks
- Character count
- Word count
- Content age
- Days since last content update

Missingness indicators were added for metadata fields where missing values could be informative.

### Exclusions

The following were excluded from model inputs:

- June impressions
- June clicks
- June CTR
- June average position
- The calculated April-to-June impression change
- The decline label
- Client identifiers as predictive features
- Content identifiers as predictive features

Client and content identifiers were used only for joins, grouping, and the client-grouped holdout split.

No client names, domains, URLs, private queries, credentials, raw exports, or other private information are published.

### Data limitations

GSC history is not equally available for every client. The warehouse contains 104 clients, with observed GSC start dates ranging from January 27, 2025 to June 2, 2026. Therefore, a common global history cannot be assumed for every client.

The analysis uses only observations where the required GSC data were available.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## 3. Methodology

### Task framing

The capstone is framed as Refresh / Content Opportunity Scoring.

The practical goal is to rank content items for human review rather than automatically prescribe a content change.

### Outcome definition

The observed outcome is a binary decline indicator.

A content item is labeled `1` when its June impressions are at least 20% lower than its April impressions:

`impression_change_pct <= -20`

The outcome is calculated only after the historical feature set has been established.

Among the final 104,301 modeling records, 77,431 were labeled declining and 26,870 were not declining.

This label is an observed search-performance outcome. It is not a measure of Google's internal ranking state and should not be interpreted as causal evidence.

### Feature design

The model uses information available before the June outcome:

- April impressions
- April clicks
- April CTR
- April average position
- Search volume
- Competition
- Backlinks
- Character count
- Word count
- Content age
- Days since update
- Missingness indicators for selected metadata fields

Features derived from June outcomes were deliberately excluded.

### Leakage checks

The final model feature matrix was checked to ensure that outcome-derived fields were not included.

Excluded fields included:

- `impressions_jun`
- `clicks_jun`
- `ctr_jun`
- `position_jun`
- `impression_change_pct`
- `is_declining`

The deliberate separation between historical inputs and the later observed outcome prevents direct label leakage.

### Baseline

A transparent rule-based baseline was constructed from three historical opportunity signals:

1. Higher April impressions receive a higher opportunity rank.
2. Lower April CTR receives a higher opportunity rank.
3. Greater time since the last content update receives a higher opportunity rank.

The final baseline score was calculated as:

- 40% April impression rank
- 35% CTR opportunity rank
- 25% freshness opportunity rank

The baseline is intentionally simple and interpretable.

### ML models

Two supervised classification models were tested:

- Logistic Regression
- Random Forest

Logistic Regression was selected as an interpretable linear model. Random Forest was added to test whether nonlinear relationships improved prioritization.

### Validation design

A client-grouped holdout split was used.

80% of the records were used for training and 20% were held out for evaluation. Complete clients were assigned to either training or holdout, rather than splitting individual rows randomly.

The resulting split contained:

- 99,796 training records
- 4,505 holdout records
- 35 training clients
- 9 holdout clients

The measured client overlap between training and holdout was zero.

### Evaluation metric

The primary metric is Precision@50.

This metric reflects the practical decision being supported: how many of the first 50 recommendations are actually observed as declining in the held-out data.

All approaches were compared on the same client-grouped holdout.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

## 4. Results (vs baseline)

### Model comparison

All three approaches were evaluated on the same client-grouped holdout set.

| Approach | Precision@50 |
|---|---:|
| Rule baseline | 0.80 |
| Logistic Regression | 0.56 |
| Random Forest | 0.68 |

### Main finding

The rule-based baseline achieved the highest measured Precision@50 at 0.80.

Random Forest achieved 0.68, which was 0.12 lower than the baseline.

Logistic Regression achieved 0.56, which was 0.24 lower than the baseline.

Therefore, in this experiment, the additional ML complexity did not improve top-50 prioritization over the simple rule-based baseline.

### Interpretation

The result suggests that the historical signals used by the baseline already captured a strong amount of the observable prioritization signal in this dataset and evaluation window.

The models should not be described as failures or as universally inferior to rules. The result is limited to the selected features, target definition, time window, client-grouped holdout, and tested model configurations.

### Holdout interpretation

The baseline identified 40 declining items among its top 50 held-out recommendations.

Random Forest identified 34 declining items among its top 50.

Logistic Regression identified 28 declining items among its top 50.

These are measured results on one holdout split and should be treated as directional evidence for decision support rather than a guarantee of future performance.

## 5. Limitations

*What this work cannot claim.*

## 5. Limitations

This analysis has several important limitations.

First, the decline label is a defined proxy based on observed impression movement between April and June. It is not Google's internal ranking signal and cannot explain why a page changed position or visibility.

Second, the analysis covers one selected time window. Search behavior can change over time because of seasonality, competition, demand changes, technical issues, or other factors.

Third, the evaluation uses one client-grouped holdout split. The observed Precision@50 values therefore do not guarantee the same performance on future clients or future time periods.

Fourth, the content opportunity score is a prioritization mechanism rather than a causal model. A high score does not prove that refreshing a page will increase rankings, clicks, or traffic.

Fifth, metadata coverage is uneven. Some fields, including backlinks, word count, character count, search volume, and optimization dates, contain missing values.

Sixth, the analysis intentionally excludes private client details, URLs, private queries, and other identifying information. This protects confidentiality but also limits the level of operational detail that can be shown publicly.

Finally, the comparison shows that a simple baseline outperformed both tested ML models on this holdout. The study therefore does not support a claim that machine learning is always superior to simple rules for refresh prioritization.

### Reason codes

The ranked queue uses simple human-readable reason codes:

- `LOW_CTR_OPPORTUNITY` — the content had relatively low historical CTR.

### Recommended workflow

High-priority items should receive human review first.

The reviewer should confirm:

1. The content is still strategically relevant.
2. The observed decline is not caused by a temporary or known external event.
3. Search demand still exists.
4. The page has a realistic opportunity for improvement.
5. A refresh is preferable to monitoring, consolidation, or no action.

The ranking is therefore an editorial prioritization aid, not an automatic rewrite instruction.
