# 1. Ranked actions + reason codes

The action queue uses the previously generated refresh output. Items are ordered using the existing `final_rank` and `final_refresh_score`.

Each recommendation contains a suggested action, confidence level, and reason codes. These signals explain why an item was prioritized.

The recommendations are intended for human review. A high score does not guarantee that changing a page will improve traffic, rankings, or conversions.

```python
# Keep the existing model-derived ranking.
queue = queue.sort_values("final_rank").reset_index(drop=True)

# Translate common reason codes into plain-language explanations.
reason_labels = {
    "declining_with_demand": "Demand is declining",
    "low_ctr_visible_page": "Low CTR despite search visibility",
    "model_decline_risk": "Model indicates decline risk",
    "moderate_ctr_visible_page": "Moderate CTR with search visibility",
    "low_impressions": "Low search visibility",
    "stale_content": "Content appears stale",
    "old_content": "Older content",
    "low_engagement": "Low engagement",
    "position_opportunity": "Ranking position offers improvement opportunity",
}

def explain_reason_codes(value):
    if pd.isna(value):
        return "No reason code recorded"

    codes = str(value).split("|")

    explanations = [
        reason_labels.get(
            code,
            code.replace("_", " ")
        )
        for code in codes
    ]

    return "; ".join(explanations)

queue["reason_explanation"] = queue["final_reason_codes"].apply(
    explain_reason_codes
)

# Create the final human-readable action queue.
action_queue = queue[
    [
        "final_rank",
        "content_id",
        "client_id",
        "final_refresh_score",
        "confidence",
        "suggested_action",
        "final_reason_codes",
        "reason_explanation",
        "trend_direction",
        "ctr",
        "avg_position",
        "content_age_days",
        "days_since_last_update",
        "content_type",
        "main_intent",
        "freshness_tier",
        "impression_tier",
        "position_tier",
    ]
].copy()

action_queue = action_queue.rename(
    columns={
        "final_rank": "rank",
        "final_refresh_score": "priority_score",
        "suggested_action": "recommended_action",
        "final_reason_codes": "reason_codes",
    }
)

print("Top 20 ranked actions:")
display(action_queue.head(20))
```

# ML-10 — Content Action Playbook

This notebook converts a validated content-refresh scoring output into a practical,
human-reviewed content action playbook.

The output is intended for decision-support: it ranks content that may deserve
refresh or review based on observed content, traffic, freshness, trend, and model
signals.

The recommendations are directional rather than production decisions. A human
should review the underlying page and business context before taking action.

## 1. Ranked actions + reason codes

The ranked queue contains 200 recommendations ordered by the final refresh score.

The highest-ranked records receive the strongest priority because they combine
multiple observed signals associated with possible content refresh opportunities.

Reason codes make the recommendations easier to interpret. Examples include
declining demand, low CTR on visible pages, model-indicated decline risk, and
low engagement on visible pages.

The queue is used as a prioritization aid rather than as an automatic publishing
decision.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Load the previously generated refresh queue
queue_path = "outputs/refresh_queue_sample.csv"

queue = pd.read_csv(queue_path)

print("Queue shape:", queue.shape)
print("Queue columns:")
print(queue.columns.tolist())

display(queue.head(10))
```

```python
# Create the ML-10 action queue from the validated refresh queue

action_queue = queue.copy()

# Rename the existing queue columns to ML-10 names
action_queue = action_queue.rename(columns={
    "final_rank": "rank",
    "final_refresh_score": "priority_score",
    "final_reason_codes": "reason_codes",
    "suggested_action": "recommended_action"
})

# Create human-readable explanations from the reason codes
def explain_reasons(row):
    codes = str(row.get("reason_codes", ""))

    explanations = []

    if "declining_with_demand" in codes:
        explanations.append("Demand is declining")

    if "low_ctr_visible_page" in codes:
        explanations.append("CTR is low despite search visibility")

    if "low_engagement_visible_page" in codes:
        explanations.append("Engagement is low despite visibility")

    if "model_decline_risk" in codes:
        explanations.append("Model indicates decline risk")

    if not explanations:
        explanations.append(
            "Multiple observed signals contributed to prioritization"
        )

    return "; ".join(explanations)


action_queue["reason_explanation"] = action_queue.apply(
    explain_reasons,
    axis=1
)

print("Top 20 ranked actions:")

display(
    action_queue[
        [
            "rank",
            "content_id",
            "priority_score",
            "confidence",
            "recommended_action",
            "reason_codes",
            "reason_explanation",
            "trend_direction",
            "ctr",
            "avg_position",
            "content_age_days",
            "days_since_last_update",
            "content_type",
            "main_intent",
            "freshness_tier",
            "impression_tier",
            "position_tier"
        ]
    ].head(20)
)
```

## 2. Intended use and limits

### Intended use

The playbook is intended for SEO/content teams to prioritize which existing
content should be reviewed first.

A reviewer can use the ranked queue to:

1. identify higher-priority content,
2. understand the signals behind the recommendation,
3. inspect the page and its current search intent,
4. decide whether a refresh is appropriate,
5. record the final human decision.

### Limits

The score does not prove that refreshing a page will improve performance.

The model was evaluated on the available dataset and should be treated as
decision-support rather than a production system.

The recommendations may become less reliable when search behavior, website
structure, content strategy, tracking, or the underlying data distribution
changes.

Missing values are present in several source fields, so recommendations should
not be interpreted as complete information about every page.

The queue also does not contain the full editorial, legal, brand, or business
context needed to make a final content decision.

## 3. Human review + the no-go list

### Human review rules

Every recommendation should be reviewed by a person before action.

The reviewer should check:

- whether the page still matches the intended search intent;
- whether the topic is still relevant;
- whether the information is accurate and current;
- whether the page has important business or brand context;
- whether the observed decline could be caused by tracking or technical issues;
- whether competitors or search-result changes explain the movement;
- whether the proposed refresh is proportionate to the opportunity;
- whether the page should actually be consolidated, redirected, left unchanged,
  or reviewed for another reason.

### Confidence handling

High-confidence recommendations can be reviewed earlier, but confidence does not
mean that the recommendation is automatically correct.

Medium-confidence recommendations should receive additional scrutiny before
action.

### No-go list

The following should NOT be automated:

- automatic publishing of rewritten content;
- automatic deletion of pages;
- automatic redirects or canonical changes;
- automatic changes to important business claims;
- automatic changes to regulated, legal, medical, financial, or safety-sensitive
  content;
- automatic changes to brand messaging;
- automatic keyword targeting without editorial review;
- automatic interpretation of model scores as guaranteed traffic or revenue gains.

## 4. Monitoring / retrain triggers

The playbook should be monitored because content performance and search behavior
can change over time.

### Monitoring signals

The following should be checked periodically:

- distribution of recommendation scores;
- distribution of confidence levels;
- proportion of each recommended action;
- trend-direction distribution;
- CTR and average-position distributions;
- missing-value rates;
- feature distributions compared with the training data;
- performance of the underlying decline/refresh model when new labeled outcomes
  become available.

### Possible retrain triggers

A model review or retraining should be considered when:

1. the input data distribution changes substantially;
2. important feature missingness increases;
3. recommendation distributions change unexpectedly;
4. model performance declines on newly labeled validation data;
5. search behavior or content strategy changes materially;
6. the definition of the target outcome changes.

These are monitoring and review triggers, not automatic retraining commands.

### Observed monitoring snapshot

The current 200-row action queue is dominated by declining-trend content: 98.5%
of the queue has a downward trend direction.

The queue contains 80.5% high-confidence recommendations and 19.5% medium-confidence
recommendations. The median priority score is 77.57, while the median CTR is 0.11
and the median average position is 11.65.

The source dataset also contains missing values. The highest observed missingness
is in `char_count` and `word_count`, both at 25.66%. `trend_pct` has 11.29%
missingness, while `scroll_rate` has 0.42%.

These measurements should be treated as a baseline monitoring snapshot. They do
not by themselves prove model degradation or business impact.

### Monitoring and retrain decision rules

The monitoring snapshot should be compared with future runs of the same pipeline.

A review should be triggered if:

- the proportion of declining recommendations changes substantially;
- the confidence distribution shifts unexpectedly;
- priority-score distributions change substantially;
- important input features develop materially higher missingness;
- the distribution of recommended actions changes unexpectedly;
- newly available labeled outcomes show weaker model performance;
- the underlying search or content environment changes materially.

Retraining should only be considered after investigating the cause of the change.
A change in one monitoring metric is not sufficient evidence by itself.

The purpose of these triggers is to identify when the current recommendations may
no longer represent the observed data conditions.

```python
import nbformat as nbf
from pathlib import Path

notebook_path = Path("work/notebooks/w07_action_playbook.ipynb")

nb = nbf.v4.new_notebook()

cells = []

# ============================================================
# TITLE
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
# ML-10 — Content Action Playbook

## Purpose

This notebook converts the validated content-refresh model output into a practical,
human-reviewed content action queue.

The output is intended as **decision-support**, not autonomous content publishing.
Recommendations are directional and should be reviewed by a human before action.

The analysis uses the anonymized internship dataset and previously validated model
outputs. No client names, URLs, private queries, titles, or keywords are used.
"""))

# ============================================================
# SECTION 1
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## 1. Ranked actions + reason codes

The ranked queue prioritizes pages for review using the validated refresh score and
associated reason codes.

The main observed action patterns are:

- **refresh_and_review_ctr** — prioritize pages showing declining demand and weak CTR
  despite having visible search performance.
- **refresh** — prioritize pages where the model indicates elevated decline risk.
- **refresh_and_review_engagement** — prioritize pages where declining demand is
  combined with weaker engagement signals.

The queue is a prioritization tool. A higher score means that the page should be
reviewed earlier; it does **not** mean that a refresh is guaranteed to improve
performance.
"""))

cells.append(nbf.v4.new_code_cell("""
import pandas as pd
from pathlib import Path

# Repository paths
queue_path = Path("work/outputs/ml10_ranked_action_queue.csv")

# Load the generated ML-10 queue
action_queue = pd.read_csv(queue_path)

print("Queue rows:", len(action_queue))
print("Queue columns:")
print(action_queue.columns.tolist())

# Support both the ML-10 naming convention and the earlier queue naming convention.
rename_map = {
    "final_rank": "rank",
    "final_refresh_score": "priority_score",
    "suggested_action": "recommended_action",
    "final_reason_codes": "reason_codes",
    "best_model_probability": "model_probability"
}

action_queue = action_queue.rename(
    columns={k: v for k, v in rename_map.items() if k in action_queue.columns}
)

# Make sure the queue is ranked correctly.
if "rank" in action_queue.columns:
    action_queue = action_queue.sort_values("rank").reset_index(drop=True)

display_cols = [
    c for c in [
        "rank",
        "content_id",
        "priority_score",
        "confidence",
        "recommended_action",
        "reason_codes",
        "reason_explanation",
        "trend_direction",
        "ctr",
        "avg_position",
        "content_age_days",
        "days_since_last_update",
        "content_type",
        "main_intent",
        "freshness_tier",
        "impression_tier",
        "position_tier"
    ]
    if c in action_queue.columns
]

print("Top 20 ranked actions:")
display(action_queue[display_cols].head(20))

print("\\nTotal recommendations:", len(action_queue))
"""))

# ============================================================
# SECTION 2
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## 2. Intended use and limits

### Intended use

This playbook is intended for a content or SEO analyst who needs to decide which
existing pages should be reviewed first.

The workflow can be used to:

1. Rank content opportunities.
2. Identify the main observed reason for prioritization.
3. Direct human attention toward pages with stronger refresh signals.
4. Support a repeatable content-review process.
5. Provide traceable evidence for the recommendations section of the research paper.

### Important limits

This is **decision-support**, not an automatic content optimizer.

The score does not prove that a page will lose traffic or that refreshing it will
increase traffic.

The analysis is based on the available anonymized dataset and measured historical
signals. Missing values are present in several features, including word count,
character count, trend percentage, search volume, competition, CPC, and main intent.

The queue should therefore not be interpreted as a causal estimate of the value of
a refresh.

The model should not automatically rewrite, delete, redirect, publish, or unpublish
content.

Recommendations should be re-evaluated when the underlying data, search behavior,
content strategy, or model changes.
"""))

cells.append(nbf.v4.new_code_cell("""
# Dataset-level context and queue statistics

data_path = Path("data/raw/content_refresh_anonymized.csv")
df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)

# Missingness among important analytical features
important_features = [
    "word_count",
    "char_count",
    "trend_pct",
    "search_volume",
    "competition",
    "cpc",
    "main_intent",
    "scroll_rate"
]

missing_summary = (
    df[important_features]
    .isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .rename("missing_pct")
    .reset_index()
    .rename(columns={"index": "feature"})
)

print("Missingness in important features:")
display(missing_summary)

# Queue-level summary
summary = {
    "queue_rows": int(len(action_queue)),
    "high_confidence_pct": round(
        (action_queue["confidence"].astype(str).str.lower() == "high").mean() * 100, 1
    ) if "confidence" in action_queue.columns else None,
    "medium_confidence_pct": round(
        (action_queue["confidence"].astype(str).str.lower() == "medium").mean() * 100, 1
    ) if "confidence" in action_queue.columns else None,
    "downtrend_pct": round(
        (action_queue["trend_direction"].astype(str).str.lower() == "down").mean() * 100, 1
    ) if "trend_direction" in action_queue.columns else None,
    "median_priority_score": round(
        action_queue["priority_score"].median(), 3
    ) if "priority_score" in action_queue.columns else None,
    "median_ctr": round(
        action_queue["ctr"].median(), 3
    ) if "ctr" in action_queue.columns else None,
    "median_avg_position": round(
        action_queue["avg_position"].median(), 3
    ) if "avg_position" in action_queue.columns else None
}

print("\\nQueue monitoring snapshot:")
print(summary)
"""))

# ============================================================
# SECTION 3
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## 3. Human review + the no-go list

### Required human review

Before acting on any recommendation, a reviewer should check:

1. **Search demand** — confirm that the observed demand trend is meaningful rather
   than a short-term fluctuation.
2. **CTR and ranking** — check whether weak CTR is consistent with the page's
   current search position.
3. **Content age/freshness** — verify that the page is actually old or stale enough
   to justify review.
4. **Search intent** — confirm that the existing content still matches the likely
   user intent.
5. **Content quality** — inspect accuracy, completeness, clarity, and usefulness.
6. **Business context** — check whether the content has strategic or commercial
   importance that is not represented by the model.
7. **Recent changes** — verify that the model is not reacting to a change that has
   already been addressed.

### No-go list

The system should **not** automatically:

- Publish rewritten content.
- Delete content.
- Redirect URLs.
- Change important business claims.
- Change legal, medical, financial, or safety-sensitive information.
- Decide that content is factually wrong.
- Decide that a ranking change was caused by a particular factor.
- Override editorial or business judgment.
- Treat model confidence as approval for publication.
- Make irreversible SEO changes without human review.

The safest interpretation is:

> **The model decides what deserves attention first; a human decides what action is appropriate.**
"""))

cells.append(nbf.v4.new_code_cell("""
# Review workload by recommendation and confidence

if "recommended_action" in action_queue.columns:
    action_distribution = (
        action_queue["recommended_action"]
        .value_counts()
        .rename_axis("recommended_action")
        .reset_index(name="count")
    )

    print("Recommended action distribution:")
    display(action_distribution)

if "confidence" in action_queue.columns:
    confidence_distribution = (
        action_queue["confidence"]
        .value_counts()
        .rename_axis("confidence")
        .reset_index(name="count")
    )

    print("Confidence distribution:")
    display(confidence_distribution)
"""))

# ============================================================
# SECTION 4
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## 4. Monitoring / retrain triggers

The playbook should be monitored rather than treated as permanently valid.

### Recommended monitoring signals

Monitor:

- Distribution of priority scores.
- Distribution of confidence levels.
- Percentage of recommendations associated with downtrends.
- Recommendation/action mix.
- Feature missingness.
- Model performance on newly observed outcomes.
- Differences between model recommendations and the human review outcome.

### Suggested retrain/review triggers

A model review should be considered when:

1. Important feature missingness increases materially compared with the current
   baseline.
2. The distribution of priority scores changes substantially.
3. The proportion of high-confidence recommendations changes unexpectedly.
4. Newly observed outcomes show materially weaker ranking performance than the
   validation result.
5. Search behavior or the underlying content population changes substantially.
6. The meaning or availability of important input features changes.

These are monitoring triggers, not automatic retraining commands. A human should
confirm the cause before retraining.
"""))

cells.append(nbf.v4.new_code_cell("""
# Lightweight monitoring snapshot

monitoring_snapshot = {
    "queue_rows": int(len(action_queue)),
    "high_confidence_pct": round(
        (action_queue["confidence"].astype(str).str.lower() == "high").mean() * 100, 2
    ),
    "medium_confidence_pct": round(
        (action_queue["confidence"].astype(str).str.lower() == "medium").mean() * 100, 2
    ),
    "downtrend_pct": round(
        (action_queue["trend_direction"].astype(str).str.lower() == "down").mean() * 100, 2
    ),
    "median_priority_score": round(action_queue["priority_score"].median(), 4),
    "mean_priority_score": round(action_queue["priority_score"].mean(), 4),
    "median_ctr": round(action_queue["ctr"].median(), 4),
    "median_avg_position": round(action_queue["avg_position"].median(), 4)
}

print("Monitoring snapshot:")
display(pd.DataFrame([monitoring_snapshot]))
"""))

# ============================================================
# SECTION 5
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## 5. Exports for the paper

The ranked queue and monitoring summaries are exported to `work/outputs/`.

The queue CSV is treated as a generated artifact rather than a source file that
must be committed. The notebook contains the code required to regenerate it.

The exports provide traceable inputs for the recommendations and monitoring
sections of the research paper.
"""))

cells.append(nbf.v4.new_code_cell("""
import json

output_dir = Path("work/outputs")
output_dir.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Ranked action queue
# ------------------------------------------------------------

ranked_queue_path = output_dir / "ml10_ranked_action_queue.csv"

action_queue.to_csv(
    ranked_queue_path,
    index=False
)

# ------------------------------------------------------------
# 2. Action distribution
# ------------------------------------------------------------

action_distribution = (
    action_queue["recommended_action"]
    .value_counts()
    .rename_axis("recommended_action")
    .reset_index(name="count")
)

action_distribution_path = output_dir / "ml10_action_distribution.csv"

action_distribution.to_csv(
    action_distribution_path,
    index=False
)

# ------------------------------------------------------------
# 3. Monitoring snapshot
# ------------------------------------------------------------

monitoring_path = output_dir / "ml10_monitoring_snapshot.json"

with open(monitoring_path, "w", encoding="utf-8") as f:
    json.dump(
        monitoring_snapshot,
        f,
        indent=2
    )

print("Exports created successfully:")
print(" -", ranked_queue_path)
print(" -", action_distribution_path)
print(" -", monitoring_path)

print("\\nAction distribution:")
display(action_distribution)
"""))

# ============================================================
# SELF CHECK
# ============================================================

cells.append(nbf.v4.new_markdown_cell("""
## Self-check

- [x] Ranked actions and reason codes are generated from the ML-10 queue.
- [x] Intended use and limitations are documented.
- [x] Human-review requirements and no-go cases are documented.
- [x] Monitoring and retrain/review triggers are documented.
- [x] Queue and monitoring exports are generated under `work/outputs/`.
- [x] Claims are framed as observed, measured, directional, or decision-support.
- [x] No client names, URLs, private queries, or private content are included.
- [ ] Notebook has been run top-to-bottom successfully in the final environment.
- [ ] Notebook has been committed under `work/notebooks/`.
- [ ] Repository URL has been submitted on the ML-10 assignment card.

### Final interpretation

This playbook should be used to prioritize human review of existing content.
It does not establish causality and does not authorize automatic publishing,
rewriting, deletion, or other irreversible content changes.
"""))

nb["cells"] = cells
nb["metadata"] = {
    "kernelspec": {
        "display_name": "Python 3",
        "language": "python",
        "name": "python3"
    },
    "language_info": {
        "name": "python"
    }
}

with open(notebook_path, "w", encoding="utf-8") as f:
    nbf.write(nb, f)

print("Completed notebook written to:")
print(notebook_path)
print("Number of cells:", len(nb["cells"]))
```
